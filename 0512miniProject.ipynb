```python
df = df.drop('ad_channel', axis=1)
```

```python
from sklearn.model_selection import train_test_split
# 필요한 데이터 가져오기
X = df[['broadcast_time', 'discount_rate', 'host_popularity']]
y = df['sales']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size = 0.3,
    random_state = 42
)

print(X_train.shape)
```

```python
# 1. 다중선형회귀 모델 만들기
from sklearn.linear_model import LinearRegression
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)

print("기울기", linear_model.coef_)
print("절편", linear_model.intercept_)
```

```python
# 여기서 확인한 것은 방송시간이 는다고 해서 판매량 증가에는 큰 도움이 되지 않고
# 오히려 약간 감소한다는 것을 확인할 수 있음 -> -0.07..
# 할인율과 진행자 인기도는 올랄갈수록 약간의 도움이 되는 것 역시 확인
```

```python
# 평가
y_pred = linear_model.predict(X_test)
print(y_pred) # 예측값
print(y_test) # 실제값

from sklearn.metrics import mean_squared_error, r2_score

# MSE : 실제값과 예측값의 차이의 제곱으로 평균을 낸 값
mse = mean_squared_error(y_test, y_pred)
print("mse:", mse)

# RMSE : 예측값이 평균 얼마나 오류가 날 수 있는지 알려줌
import numpy as np
rmse = np.sqrt(mse)
print("rmse:", rmse)

# R2 : 모델이 얼마나 정확한지 판단
r2 = r2_score(y_test, y_pred)
print("r2 score:", r2)

# R^2값이 0.987 즉 98.7% 설명할 수 있고
# RMSE는 3.67로 예측 오차 역시 낮기 때문에 성능이 좋은 모델로 학습 완료
```

```python
# 예측
new_X = [[105, 50, 89],# 방송시간도 많고 할인율도 좋고 인기도 높을때
         [60, 30, 40], # 방송시간, 할인율, 인기도가 적당할 때
         [25, 10, 19]] # 방송시간, 할인율, 인기도가 적을 때

new_y = linear_model.predict(new_X)
print(new_y)
```

```python
# 그렇다면 이젠 다항회귀 모델로 거듭나보기
# 모델 만들기
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline

poly_model = make_pipeline(
    PolynomialFeatures(degree = 2),
    LinearRegression()
)
```

```python
# 학습
poly_model.fit(X_train, y_train)
```

```python
## 계수랑 절편 확인
## 입력한 피처 그대로 보긴 힘들다.
lr = poly_model.named_steps['linearregression']
print(lr.coef_)
print(lr.intercept_)
```

```python
# 모델평가
y_pred = poly_model.predict(X_test)

print("MSE : ", mean_squared_error(y_test, y_pred))
print("RMSE : ", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R2 :",r2_score(y_test, y_pred))

# 다항회귀 모델이 다중선형회귀 모델보다 훨씬 우수하고 좋은 모델
# 다항회귀 R^2 : 99.7% || 다중선형회귀 R^2 : 98.3%
# 다항회귀 RMSE : 1.46 || 다중선형회귀 RMSE : 3.64
```

```python
# 예측
new_data = pd.DataFrame({
    "broadcast_time" : [20, 50, 80],
    "discount_rate" : [15, 30, 50],
    "host_popularity" : [25, 50, 89]
})

pred = poly_model.predict(new_data)
print(pred)
```

```python
plt.figure(figsize=(8,6))

y_pred_linear = linear_model.predict(X_test)
y_pred_poly = poly_model.predict(X_test)

# 실제값
plt.scatter(
    y_test,
    y_pred_linear,
    label="Linear Regression",
    alpha=0.7
)

plt.scatter(
    y_test,
    y_pred_poly,
    label="Polynomial Regression",
    alpha=0.7
)

# 완벽예측선
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    '--'
)

plt.xlabel("Actual Sales")
plt.ylabel("Predicted Sales")
plt.title("Linear vs Polynomial")
plt.legend()

plt.show()
```
